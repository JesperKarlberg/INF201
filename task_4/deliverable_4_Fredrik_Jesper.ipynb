```python
from Temperature import Temperature

abs_zero = Temperature(-273.15, "C")

print("Converting absolute zero to Fahrenheit and Kelvin")
print(abs_zero.to_fahrenheit())
print(abs_zero.to_kelvin())
print("\n")


cls_frht_crossover = Temperature(-40, "C")

print("Converting Celsius-Fahrenheit crossover to Fahrenheit and Kelvin")
print(cls_frht_crossover.to_fahrenheit())
print(cls_frht_crossover.to_kelvin())
print("\n")


frz_pt_wtr = Temperature(0, "C")

print("Converting freezing point water to Fahrenheit and Kelvin")
print(frz_pt_wtr.to_fahrenheit())
print(frz_pt_wtr.to_kelvin())
print("\n")


hmn_bdy_tmp = Temperature(37, "C")

print("Converting human body temperature to Fahrenheit and Kelvin")
print(hmn_bdy_tmp.to_fahrenheit())
print(hmn_bdy_tmp.to_kelvin())
print("\n")


blng_pt_wtr = Temperature(100, "C")

print("Converting boiling point water to Fahrenheit and Kelvin")
print(blng_pt_wtr.to_fahrenheit())
print(blng_pt_wtr.to_kelvin())
print("\n")


print("Testing potential conversion loss")
print(blng_pt_wtr.to_celsius()) 
```
